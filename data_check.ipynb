```python
import json
```

**Check: App Full Info & App Full Info (Adjusted) (NDJSON version)**

```python
# Specify file path
path = r"data\staging\st_app_full_info_adjusted_1766474151.ndjson"
```

```python
# Specify the number of first rows in the file to read
n = 10
```

```python
# Print out the key fields of the first N rows in the file
with open(path, "r", encoding="utf-8") as f:
    for i, line in enumerate(f):
        obj = json.loads(line)
        print(obj.get("app_id"), obj.get("os"), obj.get("name"), obj.get("unified_app_id"))
        if i == n:
            break
```

```python
# Print out all fields of a certain app_id
target_app_id = "1597222643"
with open(path, "r", encoding="utf-8") as f:
    for line in f:
        obj = json.loads(line)
        if str(obj.get("app_id")) == str(target_app_id):
            print(json.dumps(obj, ensure_ascii=False, indent=2))
            break
```
